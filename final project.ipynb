# SOCIAL MEDIA IMPACT ON STUDENTS LIFE 

```python
# 1. IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Make charts look clean

sns.set_theme(style="whitegrid")
```

```python
# 2. LOAD THE DATASET

df = pd.read_csv("Social_media_impact_on_life.csv")
print("Dataset Shape:")
print(df.shape)
print(df.head())
```

```python
# Show basic information

print("\nDataset Information:")
df.info()
```

```python
# 3.DATA CLEANING AND QUALITY CHECKING

# check missing values
print("\nMissing Values:\n")
print(df.isnull().sum())
```

```python
# check duplicate rows
print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())
```

```python
#checkunique student IDs
print("\nNumber of Unique Student IDs:")
print(df["Student_ID"].nunique())

print("\nTotal Number of Rows:")
print(len(df))
```

```python
# check categorical values
print("\nGender:\n")
print(df["Gender"].value_counts())

print("\nAcademic Level:\n")
print(df["Academic_Level"].value_counts())

print("\nMost Used Platform:\n")
print(df["Most_Used_Platform"].value_counts())

print("\nAcademic Performance Impact:\n")
print(df["Affects_Academic_Performance"].value_counts())

print("\nOverall Impact:\n")
print(df["Overall_Impact"].value_counts())
```

```python
# check numerical ranges
print("\nNumerical ranges:")
print(
    df.select_dtypes(include="number")
      .agg(["min", "max"])
      .T
)
```

```python
## 5. CREATE CLEAN DATASET
# Create a copy of the original dataset
df_clean = df.copy()
```

```python
## 6. UNIVARIATE ANALYSIS

#6.1 .Daily Social Media Usage

plt.figure(figsize=(8, 5))

sns.histplot(
    df["Avg_Daily_Usage_Hours"],
    bins=10,
    kde=True
)

plt.title("Distribution of Daily Social Media Usage")
plt.xlabel("Average Daily Usage Hours")
plt.ylabel("Number of Students")
plt.show()
```

```python
#6.2 .Sleep Hours
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Sleep_Hours_Per_Night"],
    bins=10,
    kde=True
)

plt.title("Distribution of Sleep Hours")
plt.xlabel("Sleep Hours Per Night")
plt.ylabel("Number of Students")
plt.show()
```

```python
#6.3 Mental Health Score
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Mental_Health_Score"],
    bins=10,
    kde=True
)

plt.title("Distribution of Mental Health Score")
plt.xlabel("Mental Health Score")
plt.ylabel("Number of Students")
plt.show()
```

```python
#6.4 Most Used Platforms
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="Most_Used_Platform",
    order=df["Most_Used_Platform"].value_counts().index
)

plt.title("Most Used Social Media Platforms")
plt.xlabel("Number of Students")
plt.ylabel("Platform")
plt.show()
```

```python
#6.5 Academic Performance Impact
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Affects_Academic_Performance"
)

plt.title("Does Social Media Affect Academic Performance?")
plt.xlabel("Affects Academic Performance")
plt.ylabel("Number of Students")
plt.show()
```

```python
#6.5 Overall Impact
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Overall_Impact",
    order=["Positive", "Neutral", "Negative"]
)

plt.title("Overall Impact of Social Media")
plt.xlabel("Overall Impact")
plt.ylabel("Number of Students")
plt.show()
```

```python
# 7. NUMERICAL SUMMARY
numeric_features = [
    "Age",
    "Avg_Daily_Usage_Hours",
    "Sleep_Hours_Per_Night",
    "Mental_Health_Score"
]

display(df_clean[numeric_features].describe().T)
```

```python
## 8. Bivariate Analysis

# 8.1 Social Media Usage vs Sleep
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Avg_Daily_Usage_Hours",
    y="Sleep_Hours_Per_Night"
)

plt.title("Social Media Usage vs Sleep Hours")
plt.xlabel("Average Daily Usage Hours")
plt.ylabel("Sleep Hours Per Night")
plt.show()
```

```python
# 8.2 Social Media Usage vs Mental Health

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Avg_Daily_Usage_Hours",
    y="Mental_Health_Score"
)

plt.title("Social Media Usage vs Mental Health")
plt.xlabel("Average Daily Usage Hours")
plt.ylabel("Mental Health Score")
plt.show()
```

```python
# 8.3 Usage vs Academic Performance Impact

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Affects_Academic_Performance",
    y="Avg_Daily_Usage_Hours"
)

plt.title("Daily Social Media Usage by Academic Performance Impact")
plt.xlabel("Affects Academic Performance")
plt.ylabel("Average Daily Usage Hours")
plt.show()
```

```python
# 8.4 Usage vs Overall Impact

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Overall_Impact",
    y="Avg_Daily_Usage_Hours",
    order=["Positive", "Neutral", "Negative"]
)

plt.title("Daily Usage by Overall Social Media Impact")
plt.xlabel("Overall Impact")
plt.ylabel("Average Daily Usage Hours")
plt.show()
```

```python
# 8.5. GROUP ANALYSIS

academic_group = df.groupby(
    "Affects_Academic_Performance"
)["Avg_Daily_Usage_Hours"].agg(
    ["count", "mean", "median"]
)

print("\nUsage by Academic Performance Impact:")
print(academic_group)

```

```python
# 9.MULTIVARIATE ANALYSIS

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Avg_Daily_Usage_Hours",
    y="Mental_Health_Score",
    hue="Overall_Impact"
)

plt.title("Social Media Usage, Mental Health and Overall Impact")
plt.xlabel("Average Daily Usage Hours")
plt.ylabel("Mental Health Score")
plt.show()
```

```python
# 10. CORRELATION ANALYSIS

# Select numerical columns
numeric_columns = [
    "Age",
    "Avg_Daily_Usage_Hours",
    "Sleep_Hours_Per_Night",
    "Mental_Health_Score"
]

correlation = df[numeric_columns].corr()

print("\nCorrelation Matrix:")
print(correlation)
```

```python
# 11. CORRELATION HEATMAP


plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Numerical Variables")
plt.show()
```

```python
#12 .USAGE LEVELS

# We divide students into Low, Medium and High usage groups.

df["Usage_Level"] = pd.cut(
    df["Avg_Daily_Usage_Hours"],
    bins=[0, 3, 6, 9],
    labels=["Low", "Medium", "High"]
)


print("\nUsage Level Counts:")
print(df["Usage_Level"].value_counts())
```

```python
# 13. ANALYZE USAGE LEVEL vs SLEEP

usage_sleep = df.groupby(
    "Usage_Level",
    observed=False
)["Sleep_Hours_Per_Night"].mean()

print("\nAverage Sleep by Usage Level:")
print(usage_sleep)


plt.figure(figsize=(8, 5))

usage_sleep.plot(kind="bar")

plt.title("Average Sleep Hours by Social Media Usage Level")
plt.xlabel("Usage Level")
plt.ylabel("Average Sleep Hours")
plt.xticks(rotation=0)
plt.show()
```

```python
# 13. ANALYZE USAGE LEVEL vs MENTAL HEALTH

usage_mental = df.groupby(
    "Usage_Level",
    observed=False
)["Mental_Health_Score"].mean()

print("\nAverage Mental Health Score by Usage Level:")
print(usage_mental)


plt.figure(figsize=(8, 5))

usage_mental.plot(kind="bar")

plt.title("Average Mental Health Score by Social Media Usage Level")
plt.xlabel("Usage Level")
plt.ylabel("Average Mental Health Score")
plt.xticks(rotation=0)
plt.show()
```

```python
# 14. FEATURE EXPLORATION WITH BOXPLOTS

features = [
    "Age",
    "Avg_Daily_Usage_Hours",
    "Sleep_Hours_Per_Night",
    "Mental_Health_Score"
]

for column in features:

    plt.figure(figsize=(7, 4))

    sns.boxplot(
        y=df[column]
    )

    plt.title(f"Boxplot of {column}")
    plt.ylabel(column)

    plt.show()
```
